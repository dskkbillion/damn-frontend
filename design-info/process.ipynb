```python
import json
# import uuid # Not strictly needed if using a simple counter

# Load the dataset
file_path = 'service_updated_eng.json'
try:
    with open(file_path, 'r') as f:
        dataset = json.load(f)
except FileNotFoundError:
    print(f"Error: File not found at {file_path}")
    dataset = None

if dataset is not None:
    processed_dataset = []
    variant_id_counter = 1 # Counter for generating unique variant IDs
    option_id_counter = 1 # Counter for generating unique option IDs
    option_value_id_counter = 1 # Counter for generating unique option value IDs


    for item in dataset:
        # 1. Add state if missing
        if 'state' not in item:
            item['state'] = 'NORMAL'

        # 2. Add tenantId
        item['tenantId'] = 1

        # 3. Add winImages
        if 'winImages' not in item:
            item['winImages'] = []

        # 4. Add productMaterials
        if 'productMaterials' not in item:
            item['productMaterials'] = []

        # --- New Logic: Generate options if missing ---
        # Check if 'options' field is missing or is an empty list
        if 'options' not in item or not item['options']:
             generated_options = []
             num_options_to_generate = 3 # Generate 3 ProductOption objects
             num_values_per_option = 2 # Generate 2 ProductOptionValue objects for each option

             for i in range(num_options_to_generate):
                 option = {
                     "id": option_id_counter,
                     "tenantId": item['tenantId'],
                     "productId": item['id'],
                     "name": f"Option Type {i + 1}", # Generic English name
                     "sort": i,
                     "createTime": None, # Set to null
                     "updateTime": None, # Set to null
                     "values": [] # Will populate this list
                 }
                 option_id_counter += 1 # Increment option ID counter

                 generated_values = []
                 for j in range(num_values_per_option):
                     value = {
                         "id": option_value_id_counter,
                         "tenantId": item['tenantId'],
                         "productId": item['id'],
                         "optionId": option['id'], # Link to parent option
                         "name": f"Value {chr(65 + j)}", # Value A, Value B, etc.
                         "sort": j,
                         "createTime": None, # Set to null
                         "updateTime": None, # Set to null
                     }
                     generated_values.append(value)
                     option_value_id_counter += 1 # Increment value ID counter

                 option['values'] = generated_values
                 generated_options.append(option)

             item['options'] = generated_options
        # --- End New Logic ---


        # --- Generate and Add variants (Existing logic, runs for ALL items) ---
        # This part still runs for ALL items to ensure they have the variants field
        generated_variants = []
        base_selling_price = item.get('sellingPrice', 0)

        variant_tiers_english = [
            {"name": "Basic Tier", "price_multiplier": 0.8},
            {"name": "Standard Tier", "price_multiplier": 1.0},
            {"name": "Premium Tier", "price_multiplier": 1.2},
        ]

        for tier in variant_tiers_english:
            variant = {
                "id": variant_id_counter, # Use counter for unique ID
                "name": tier["name"],
                "tenantId": item['tenantId'],
                "productId": item['id'],
                "sellingPrice": round(base_selling_price * tier["price_multiplier"], 2), # Derived price, rounded
                "originalPrice": None, # Set to null
                "costPrice": None, # Set to null
                "inventory": 10, # Default inventory
                "deliveryDay": 3, # Default delivery day
                "editNum": 1, # Default edit num
                "image": None, # Set to null
                "code": None, # Set to null
                "feature": [], # Default empty list
                "discountAmount": None, # Set to null
                "sort": None, # Set to null
                "createTime": None, # Set to null
                "updateTime": None, # Set to null
            }
            generated_variants.append(variant)
            variant_id_counter += 1 # Increment counter

        item['variants'] = generated_variants
        # --- End variant generation ---


        processed_dataset.append(item)

    # Save the processed data to a new file
    output_file_path = 'service_processed_eng_2.json'
    with open(output_file_path, 'w') as f:
        json.dump(processed_dataset, f, indent=2)
    print(f"Processed data saved to {output_file_path}")

else:
    print("Could not process the dataset.")
```
